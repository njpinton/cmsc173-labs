## Part 3 &middot; Centring and scaling

Week 2 asks you to normalise features before gradient descent. Here is why: these
three stations are on wildly different scales, and a distance- or gradient-based
method will let the largest one dominate.

```python
raw = np.array([
    [12.0,  1200.0,  0.35],    # rainfall mm, elevation m, humidity fraction
    [ 0.0,   340.0,  0.28],
    [45.2,  2100.0,  0.71],
    [ 6.4,   890.0,  0.42],
    [22.1,  1550.0,  0.55],
])

mu = raw.mean(axis=0)
sigma = raw.std(axis=0)
z = (raw - mu) / sigma

print("column means before:", np.round(mu, 2))
print("column stds  before:", np.round(sigma, 2))
print("\ncolumn means after :", np.round(z.mean(axis=0), 6))
print("column stds  after :", np.round(z.std(axis=0), 6))
```
