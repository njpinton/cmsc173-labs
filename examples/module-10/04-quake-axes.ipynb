## Which way do the Philippines point?

PCA on 11,391 earthquake epicentres: longitude and latitude only. PC1 is the direction the quakes spread along most, and the archipelago traces it.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

url = "https://raw.githubusercontent.com/njpinton/cmsc173-labs/main/data/ph_earthquakes.csv"
geo = pd.read_csv(url, parse_dates=["time"])[["longitude", "latitude"]]
pca = PCA().fit(geo)                            # both columns are degrees: one unit, so no scaler
print(pca.components_.round(3))                 # PC1 [-0.426  0.905] (or its opposite)
print(pca.explained_variance_ratio_.round(3))   # [0.908 0.092]
pc1 = np.abs(pca.components_[0])
print("PC1 runs", round(np.degrees(np.arctan2(pc1[0], pc1[1])), 1), "degrees west of north")   # 25.2
```

```python
plt.figure(figsize=(5, 6.5))
plt.scatter(geo.longitude, geo.latitude, s=2, alpha=0.3)
m = pca.mean_
for comp, var in zip(pca.components_, pca.explained_variance_):
    d = comp * 2 * np.sqrt(var)
    plt.plot([m[0] - d[0], m[0] + d[0]], [m[1] - d[1], m[1] + d[1]], lw=3)
plt.gca().set_aspect("equal"); plt.show()
```

**Standardising here would be a mistake.** With two standardised features, PC1 is *always* the 45&deg; diagonal: the direction the data actually runs is thrown away.

```python
scaled = make_pipeline(StandardScaler(), PCA()).fit(geo)
print(scaled[-1].components_.round(3), scaled[-1].explained_variance_ratio_.round(3))   # 0.707s; [0.869 0.131]
```
