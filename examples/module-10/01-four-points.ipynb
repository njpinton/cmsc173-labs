## Four points, by hand

Four points: (1, 2), (2, 1), (3, 4), (4, 3). Centre them, measure their spread along four directions, and the first principal component is the direction with the most.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[1, 2], [2, 1], [3, 4], [4, 3]])
C = X - X.mean(axis=0)                       # centre: the mean is (2.5, 2.5)
print(C)
print("variances:", C.var(axis=0), "covariance:", (C[:, 0] * C[:, 1]).mean())   # 1.25 1.25 | 0.75
```

**Spread along a direction.** A direction is a pair $(u_1, u_2)$ of length 1. Each point's score is the weighted sum $u_1 x_1 + u_2 x_2$; the spread is the variance of the four scores.

```python
for deg in (0, 45, 90, 135):
    u = np.array([np.cos(np.radians(deg)), np.sin(np.radians(deg))])
    scores = C[:, 0] * u[0] + C[:, 1] * u[1]
    print(deg, u.round(3), scores.round(3), "variance", round(scores.var(), 3))
# 45 degrees: 2.0, the most. 135 degrees: 0.5, the least. Together 2.5 = 1.25 + 1.25.
```

```python
deg = np.linspace(0, 180, 181)
spread = [(C[:, 0] * np.cos(np.radians(d)) + C[:, 1] * np.sin(np.radians(d))).var() for d in deg]
plt.plot(deg, spread); plt.xlabel("direction (degrees)"); plt.ylabel("variance of the scores"); plt.show()
```

**Rebuild point (1, 2) from PC1 alone.** Its PC1 score times the direction, plus the mean.

```python
pc1 = np.array([0.707, 0.707])
score = C[0, 0] * pc1[0] + C[0, 1] * pc1[1]         # -1.414
rebuilt = X.mean(axis=0) + score * pc1
print(round(score, 3), rebuilt.round(2), "squared error", round(((X[0] - rebuilt) ** 2).sum(), 2))   # (1.5, 1.5), 0.5
```

**The same in scikit-learn.** `explained_variance_` divides by $n - 1$ (2.667, 0.667); the ratio is unchanged. A component may come back pointing the opposite way: same line.

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2).fit(X)
print(pca.components_.round(3))
print(pca.explained_variance_.round(3), pca.explained_variance_ratio_)   # [2.667 0.667] [0.8 0.2]
```
