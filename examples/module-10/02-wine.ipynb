## Wine: thirteen measurements, two pictures

178 wines, 13 chemical measurements in very different units. Standardise, then PCA: how much survives in two components, and do the three cultivars show up without being told?

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

wine = load_wine()
pipe = make_pipeline(StandardScaler(), PCA()).fit(wine.data)
ratio = pipe[-1].explained_variance_ratio_
print(ratio[:3].round(3), ratio[:3].sum().round(3))      # [0.362 0.192 0.111] 0.665
```

```python
plt.bar(range(1, 14), ratio); plt.xlabel("component"); plt.ylabel("share of variance"); plt.show()
```

```python
Z = pipe.transform(wine.data)                 # 178 rows, now scored on the 13 components
plt.scatter(Z[:, 0], Z[:, 1], c=wine.target, cmap="viridis", alpha=0.7)
plt.xlabel("PC1"); plt.ylabel("PC2"); plt.title("colour = cultivar, which PCA never saw"); plt.show()
```

**What each component is made of** (its *loadings*): the four largest weights, by size. A whole component may come back with every sign flipped.

```python
for k in range(2):
    w = pipe[-1].components_[k]
    top = np.argsort(-np.abs(w))[:4]
    print(f"PC{k + 1}:", [(wine.feature_names[i], round(w[i], 2)) for i in top])
```
