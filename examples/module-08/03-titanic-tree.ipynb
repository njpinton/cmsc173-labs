## The Titanic, as a tree

Module 6 fitted a logistic regression to these 714 passengers. Same features, same split: a decision tree of three questions, which you can read as rules.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")[["survived", "pclass", "sex", "age"]].dropna()
df["female"] = (df["sex"] == "female").astype(int)        # 1 = woman or girl
X, y = df[["female", "pclass", "age"]], df["survived"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=173, stratify=y)

tree = DecisionTreeClassifier(max_depth=3, random_state=173).fit(X_tr, y_tr)
print(round(tree.score(X_tr, y_tr), 3), round(tree.score(X_te, y_te), 3))   # 0.826 0.795
print(tree.feature_importances_.round(2))                                    # [0.6  0.28 0.12]
```

**The tree as rules.** `weights` are the training passengers in each leaf: [died, survived].

```python
print(export_text(tree, feature_names=list(X.columns), show_weights=True))
```

```python
plt.figure(figsize=(14, 6))
plot_tree(tree, feature_names=list(X.columns), class_names=["died", "survived"], filled=True, fontsize=9)
plt.show()
```

**Deeper is not better.** With no depth limit the tree memorises the training passengers.

```python
for depth in (1, 2, 3, 4, 5, None):
    t = DecisionTreeClassifier(max_depth=depth, random_state=173).fit(X_tr, y_tr)
    print(depth, round(t.score(X_tr, y_tr), 3), round(t.score(X_te, y_te), 3), t.get_n_leaves())
# None: 0.914 on training, 0.763 on test, 102 leaves
```
