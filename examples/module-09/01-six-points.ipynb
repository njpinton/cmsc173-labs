## Six points, by hand

The lecture's six points: three of class +1, three of class &minus;1. By hand the widest-margin line is $0.5x_1 + 0.5x_2 - 1 = 0$, fixed by A and D alone. Here scikit-learn finds the same line.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

X = np.array([[2, 2], [3, 2], [2, 3], [0, 0], [-1, 0], [0, -1]])   # points A to F
y = np.array([1, 1, 1, -1, -1, -1])
svm = SVC(kernel="linear", C=1000).fit(X, y)    # a huge C: no point may cross the margin
print(svm.coef_, svm.intercept_, svm.support_)  # [[0.5 0.5]] [-1.] [3 0]: rows 3 and 0 are D and A
print(round(2 / np.linalg.norm(svm.coef_), 3))  # margin width 2.828
```

**The score of every point**, $f(x) = 0.5x_1 + 0.5x_2 - 1$. Support vectors score exactly &plusmn;1.

```python
for name, point, label in zip("ABCDEF", X, y):
    print(name, point, label, svm.decision_function([point])[0].round(2))
```

**Only the support vectors matter.** Each gets a weight $\alpha_i y_i$ (`dual_coef_`); the line is built from them alone.

```python
print(svm.dual_coef_)                                         # [[-0.25  0.25]] for D and A
w = (svm.dual_coef_ @ svm.support_vectors_)                   # 0.25 * A - 0.25 * D
print(w)                                                      # [[0.5 0.5]]: the same w
smaller = SVC(kernel="linear", C=1000).fit(X[[0, 3]], y[[0, 3]])   # keep only A and D
print(smaller.coef_, smaller.intercept_)                      # the same line again
```

```python
xx = np.linspace(-2, 4, 2)
for level, style in ((0, "-"), (1, "--"), (-1, "--")):
    plt.plot(xx, (level + 1) / 0.5 - xx, "k" + style)        # 0.5x1 + 0.5x2 - 1 = level
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", s=80, zorder=3)
plt.scatter(*svm.support_vectors_.T, s=250, facecolors="none", edgecolors="k")
plt.axis("equal"); plt.show()
```
