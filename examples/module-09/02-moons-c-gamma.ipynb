## The moons: two dials, C and gamma

300 points in two interleaved half-moons: no straight line splits them. An RBF SVM can, and two settings decide how: $\gamma$, how far each point's influence reaches, and $C$, the cost of a violation.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X, y = make_moons(n_samples=300, noise=0.25, random_state=173)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=173)
scaler = StandardScaler().fit(X_tr)          # mean and SD from training rows only
svm = SVC(kernel="rbf", C=1.0, gamma=1.0).fit(scaler.transform(X_tr), y_tr)
print(svm.n_support_.sum(), round(svm.score(scaler.transform(X_te), y_te), 3))   # 65 0.933
```

**Turn the dials.** Training accuracy, test accuracy, support vectors.

```python
for gamma, C in ((0.1, 1), (1, 1), (100, 1), (1, 0.1), (1, 100)):
    s = SVC(kernel="rbf", C=C, gamma=gamma).fit(scaler.transform(X_tr), y_tr)
    print(f"gamma={gamma:<5} C={C:<5}", round(s.score(scaler.transform(X_tr), y_tr), 3),
          round(s.score(scaler.transform(X_te), y_te), 3), s.n_support_.sum())
# gamma=100: 1.0 on training, 0.833 on test: islands around single points
```

```python
xx, yy = np.meshgrid(np.linspace(-2.5, 2.5, 300), np.linspace(-2.5, 2.5, 300))
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
Xs = scaler.transform(X_tr)
for ax, gamma in zip(axes, (0.1, 1, 100)):
    s = SVC(kernel="rbf", gamma=gamma).fit(Xs, y_tr)
    ax.contourf(xx, yy, s.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape), alpha=0.3, cmap="coolwarm")
    ax.scatter(Xs[:, 0], Xs[:, 1], c=y_tr, cmap="coolwarm", s=15)
    ax.set_title(f"gamma = {gamma}")
plt.show()
```
