## Handwritten digits: ten classes

1,797 handwritten digits, each 8 &times; 8 = 64 pixels. Ten classes, so scikit-learn trains one SVM for every pair of digits: 45 of them.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X, y = load_digits(return_X_y=True)          # pixels 0 to 16: one shared scale, so no scaler
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=173, stratify=y)
for kernel in ("linear", "rbf"):
    svm = SVC(kernel=kernel).fit(X_tr, y_tr)
    print(kernel, round(svm.score(X_te, y_te), 3), svm.n_support_.sum())
# linear 0.987 364 | rbf 0.987 597
```

**The seven it gets wrong**, out of 540 test digits.

```python
pred = svm.predict(X_te)                      # svm is the RBF model from the loop
wrong = np.where(pred != y_te)[0]
fig, axes = plt.subplots(1, len(wrong), figsize=(12, 2.2))
for ax, i in zip(axes, wrong):
    ax.imshow(X_te[i].reshape(8, 8), cmap="gray_r")
    ax.set_title(f"a {y_te[i]}, read as {pred[i]}"); ax.axis("off")
plt.show()
print(confusion_matrix(y_te, pred))
```
