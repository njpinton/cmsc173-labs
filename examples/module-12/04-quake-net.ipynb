## How deep? Once more

Module 8 predicted earthquake depth from longitude and latitude with KNN, trees and forests. A neural network on the same split: first unscaled, then standardised.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides (scikit-learn 1.4.2; a newer version can differ slightly for the trained networks);
change something and see what moves.

```python
import warnings; warnings.filterwarnings("ignore")
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

url = "https://raw.githubusercontent.com/njpinton/cmsc173-labs/main/data/ph_earthquakes.csv"
quakes = pd.read_csv(url)
X, y = quakes[["longitude", "latitude"]], quakes["depth"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=173)
raw = MLPRegressor(hidden_layer_sizes=(64, 64), max_iter=2000, random_state=173)
scaled = make_pipeline(StandardScaler(), MLPRegressor(hidden_layer_sizes=(64, 64), max_iter=2000, random_state=173))
print(round(raw.fit(X_tr, y_tr).score(X_te, y_te), 3), round(scaled.fit(X_tr, y_tr).score(X_te, y_te), 3))   # 0.105 0.574
```

**Against module 8's models** on the same split.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor

for name, m in {"linear regression": LinearRegression(),
                "tree, depth 6": DecisionTreeRegressor(max_depth=6, random_state=173),
                "forest, 200 trees": RandomForestRegressor(200, random_state=173, n_jobs=-1),
                "KNN, k = 10": KNeighborsRegressor(10)}.items():
    print(f"{name:20s} {m.fit(X_tr, y_tr).score(X_te, y_te):.3f}")
# 0.152 | 0.621 | 0.625 | 0.651: with two features, KNN beats the network's 0.574
```
