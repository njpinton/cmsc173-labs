## Who survived the Titanic?

Module 5 explored the passengers with charts. Here a logistic regression puts numbers on what the charts showed: sex, class and age, each as an odds ratio.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

df = sns.load_dataset("titanic")[["survived", "pclass", "sex", "age"]].dropna()
df["female"] = (df["sex"] == "female").astype(int)        # 1 = woman, 0 = man
X, y = df[["female", "pclass", "age"]], df["survived"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=173, stratify=y)
model = LogisticRegression(max_iter=1000).fit(X_tr, y_tr)
print(model.coef_.round(3), model.intercept_.round(3))   # [[ 2.45  -1.305 -0.039]] [2.62]
print(round(model.score(X_tr, y_tr), 3), round(model.score(X_te, y_te), 3))   # 0.784 0.814
```

**Reading a coefficient.** The score $z$ is the log-odds, so $e^{\theta}$ is the factor the odds are multiplied by when that feature goes up by one.

```python
print({k: round(float(v), 2) for k, v in zip(X.columns, np.exp(model.coef_[0]))})   # female 11.59, pclass 0.27, age 0.96
print("ten years older:", round(np.exp(10 * model.coef_[0][2]), 2))   # 0.68
```

```python
people = pd.DataFrame({"female": [1, 0, 1, 0, 0], "pclass": [1, 1, 3, 3, 3], "age": [30, 30, 30, 30, 5]})
people["P(survived)"] = model.predict_proba(people[["female", "pclass", "age"]])[:, 1].round(2)
people        # 0.93, 0.54, 0.50, 0.08, 0.18
```
