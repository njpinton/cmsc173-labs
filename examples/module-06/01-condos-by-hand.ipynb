## Premium condos, by hand

Module 3's five condos, now with a yes/no label: is the unit *premium*? Gradient descent fits a logistic regression with nothing but NumPy.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([20, 40, 60, 80, 100.])   # floor area (sqm)
y = np.array([0, 0, 1, 0, 1.])         # 1 = premium
xs = (x - x.mean()) / x.std()          # standardise: mean 0, SD 1

theta = np.zeros(2)                    # (theta_0, theta_1)
for step in range(2000):
    p = 1 / (1 + np.exp(-(theta[0] + theta[1] * xs)))      # sigmoid of the score z
    grad = np.array([np.mean(p - y), np.mean((p - y) * xs)])
    theta -= 0.5 * grad                                      # learning rate 0.5
print(theta.round(4))                  # [-0.6227  1.5421]
```

**Back to square metres.** Divide the slope by the SD of area and correct the intercept, as in module 3.

```python
theta1 = theta[1] / x.std()
theta0 = theta[0] - theta[1] * x.mean() / x.std()
print(round(theta0, 4), round(theta1, 5))              # -3.894  0.05452
print("boundary:", round(-theta0 / theta1, 2), "sqm")  # 71.42
print("odds ratio per extra sqm:", round(np.exp(theta1), 3))
```

```python
area = np.linspace(10, 110, 200)
plt.plot(area, 1 / (1 + np.exp(-(theta0 + theta1 * area))), lw=2)
plt.scatter(x, y, s=80, color="black", zorder=3)
plt.axvline(-theta0 / theta1, ls="--", color="red")
plt.xlabel("floor area (sqm)"); plt.ylabel("P(premium)")
plt.show()
```
