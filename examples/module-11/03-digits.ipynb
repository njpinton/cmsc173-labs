## Digits without labels

1,797 handwritten digits, each 64 pixel values. K-Means with $K = 10$ is never told which digit is which; the labels are used only afterwards, to check.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_digits
from sklearn.metrics import adjusted_rand_score

X, y = load_digits(return_X_y=True)      # y is kept only to check afterwards
km = KMeans(n_clusters=10, n_init=10, random_state=173).fit(X)
print(round(adjusted_rand_score(y, km.labels_), 3))     # 0.668
```

**The ten centroids, drawn as images.** Each centroid is 64 pixel averages: reshape it to 8 &times; 8.

```python
ct = pd.crosstab(km.labels_, y)          # rows: clusters, columns: true digits
majority = ct.idxmax(axis=1)
fig, axes = plt.subplots(2, 5, figsize=(10, 4.5))
for ax, c in zip(axes.ravel(), sorted(range(10), key=lambda c: majority[c])):
    ax.imshow(km.cluster_centers_[c].reshape(8, 8), cmap="Greys")
    ax.set_title(f"mostly {majority[c]}: {ct.loc[c].max()} of {ct.loc[c].sum()}")
    ax.axis("off")
plt.show()
```

**Where the digits mix.** The three most common digits in each cluster.

```python
for c in sorted(range(10), key=lambda c: majority[c]):
    top = ct.loc[c].sort_values(ascending=False).head(3)
    print(f"mostly {majority[c]}:", dict(zip(top.index.tolist(), top.tolist())))
# mostly 1: {1: 55, 9: 20, 8: 6} | mostly 8: {8: 101, 1: 99, 2: 8} | mostly 9: {9: 139, 8: 51, 5: 43}
```
