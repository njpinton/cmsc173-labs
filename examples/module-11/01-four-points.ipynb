## K-Means on four points, by hand

Four points, A (1, 2), B (2, 1), C (4, 3) and D (5, 4), and $K = 2$. K-Means repeats two steps: each point joins its nearest centroid, then each centroid moves to the mean of its points.

**How to use this notebook.** Runtime &rarr; Run all. Every number printed below is on the
slides; change something and see what moves.

**The two steps in NumPy**, starting from A and D as the centroids.

```python
import numpy as np

X = np.array([[1, 2], [2, 1], [4, 3], [5, 4]], dtype=float)   # A, B, C, D
mu = X[[0, 3]].copy()                                          # start at A and D
for it in range(1, 11):
    dist = np.sqrt(((X[:, None, :] - mu[None, :, :]) ** 2).sum(axis=2))   # every point to every centroid
    labels = dist.argmin(axis=1)                               # assign: nearest centroid
    print("iteration", it, "distances", dist.round(2).tolist(), "labels", labels)
    new_mu = np.array([X[labels == k].mean(axis=0) for k in range(2)])     # update: the means
    if np.allclose(new_mu, mu):                                # nothing moved: stop
        break
    mu = new_mu
print("centroids", mu.tolist(), "after", it, "iterations")    # [[1.5, 1.5], [4.5, 3.5]], 2
print("inertia J =", ((X - mu[labels]) ** 2).sum())          # 2.0
```

**The same in scikit-learn.** `init=X[[0, 3]]` starts at A and D, as on the slides.

```python
from sklearn.cluster import KMeans

X = np.array([[1, 2], [2, 1], [4, 3], [5, 4]])            # A, B, C, D
km = KMeans(n_clusters=2, init=X[[0, 3]], n_init=1).fit(X)  # start at A and D
print(km.labels_, km.cluster_centers_.tolist(), km.inertia_, km.n_iter_)
# [0 0 1 1] [[1.5, 1.5], [4.5, 3.5]] 2.0 2
```

**Review question.** A new point E = (3, 2). Which cluster, and where does that centroid move?

```python
E = np.array([3, 2])
print(np.sqrt(((km.cluster_centers_ - E) ** 2).sum(axis=1)).round(2))   # [1.58 2.12]: cluster 1
print(np.array([[1, 2], [2, 1], [3, 2]]).mean(axis=0).round(2))         # [2.   1.67]
```
